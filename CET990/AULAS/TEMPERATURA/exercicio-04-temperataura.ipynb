```python
#==========================================================
# 2. Cálculo do ruído
#==========================================================

tabela["ruido_C"] = (
    tabela["sensor_ruidoso_C"] - 
    tabela["sensor_referencia_C"]
    )

tabela
```

```python
#========================================================
# 3. Caracterização quantitativa do sinal original 
#========================================================

# Média do ruído
ruido_medio = tabela["ruido_C"].mean()

# Desvio padrão amostral do ruído
desvio_ruido = tabela["ruido_C"].std()

# Erro absoluto em cada ponto de medição
tabela["erro_original_C"] = np.abs(
    tabela["ruido_C"]
    )

# Erro absoluto médio
erro_original_medio = tabela["erro_original_C"].mean()

# Apresentação dos resultaldos
print("-----------------------------------------------")
print("Caracterização do sinal original")
print("-----------------------------------------------")
print(f"Ruído médio = {ruido_medio:.4f} °C")
print(f"Desvio padrão do ruído = {desvio_ruido:.4f} °C")
print(f"Erro absoluto médio = {erro_original_medio:.4f} °C")
print("------------------------------------------------")
```
